```python
a=140
b=24

if b>a:
    print(f'{b} is greater than {a}')
else:
    print(f'{b} is less than or equal to {a}')
```

```python
a=410
b=410

if b>a:
    print(f'{b} is greater than {a}')
elif b==a:
    print(f'{b} is or equal to {a}')
else:
    print(f'{b} is less than {a}')
```

# Loops

### While

```python
i=0
while i<4:
    print(i)
    i+=1
```

```python
i = 0
while i < 10:
    print(i * 2 +3)
    i += 3
```

### For

```python
for i in range(6):
    print(i)
```

```python
for i in range(10,22,2):
    print(i)
```

```python
cars = ["Tesla","Mazda","Ford","Toyota", "Benz"]
```

```python
for car in cars: # car is variable inside the for loop
    print(car)
```

```python
for i in cars: 
    print(i)
```

```python
vowel_letters = "aeiouAEIOU"
```

```python
coursal_vowels_letters = [letter for letter in course if letter in vowel_letters]
print(coursal_vowels_letters)
```
