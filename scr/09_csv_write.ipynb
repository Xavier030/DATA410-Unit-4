```python
import csv
```

```python
names = ['Toyota', 'Ford', 'BMW']
models = ['RAV4', 'F150', 'X6']
max_speed = [200, 220, 260]
powers = [195, 400, 368]
```

```python
cars_list = list(zip(names, models, max_speed, powers))
cars_list
```

```python
with open('../datasets/cars.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['Name', 'Model', 'Max Speed', 'Power'])
    writer.writerows(cars_list)
```
